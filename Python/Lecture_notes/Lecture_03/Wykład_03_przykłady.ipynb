## return a print - slajd 15

* Który wynik można wykorzystać w dalszych obliczeniach? Co otrzymamy po przypisaniu wyniku funkcji, która tylko wypisuje wartość?


```python
def przelicz_na_V(napiecie_mV):
    return napiecie_mV / 1000

def wypisz_napiecie_V(napiecie_mV):
    print(napiecie_mV / 1000)

wynik = przelicz_na_V(1250)
print("Wartość zwrócona przez return:", wynik)
print("Wartość podwojona:", 2 * wynik)

print("Wywołanie funkcji z print: ", end='')
print(2*wypisz_napiecie_V(1250))



#wynik_wypisywania = wypisz_napiecie_V(1250)
#print("Wartość zwrócona przez funkcję z print:", wynik_wypisywania)
```

## Mutowalne wartości domyślne - slajd 17

Wartość domyślna powstaje podczas wykonywania definicji funkcji. Ponowne uruchomienie całej komórki definiuje funkcję od nowa i zeruje stan tego przykładu.

* Dlaczego drugie wywołanie pamięta poprzedni odczyt? Jak użycie `None` pozwala utworzyć osobną listę przy każdym wywołaniu bez podanej listy?

```python
def dodaj_odczyt_blednie(odczyt, odczyty=[]):
    odczyty.append(odczyt)
    return odczyty

print("Pierwsze wywołanie:", dodaj_odczyt_blednie(1.2))
print("Drugie wywołanie:", dodaj_odczyt_blednie(1.3))
```

```python
def dodaj_odczyt(odczyt, odczyty=None):
    if odczyty is None:
        odczyty = []
    odczyty.append(odczyt)
    return odczyty

print("Pierwsze wywołanie:", dodaj_odczyt(1.2))
print("Drugie wywołanie:", dodaj_odczyt(1.3))

seria = [1.0, 1.1]
dodaj_odczyt(1.2, seria)
print("Jawnie przekazana lista:", seria)
```

## Łączenie i rozpakowywanie argumentów - slajd 28

* Dokąd trafią odczyty, a dokąd informacja o operatorze? Czy jednostkę można przekazać jako kolejny argument pozycyjny?

```python
def opisz_pomiar(probka, kanal=2, *odczyty, jednostka, dokladnosc=4, **opis):
    print(f"Próbka {probka}, kanał {kanal}")
    print("Krotka odczytów:", odczyty)
    for odczyt in odczyty:
        print(f"  {odczyt:.{dokladnosc}f} {jednostka}")
    print("Dodatkowy opis:", opis)


opisz_pomiar("PR-204", 3, 1.25, 1.30, jednostka="mV", operator="Anna")
```

```python
lista = ["PR-205", 2, 0.98, 1.02]
slownik = {"jednostka": "mV", "dokladnosc": 2, "seria": "B"}

# * rozpakowuje elementy listy jako argumenty pozycyjne.
# ** rozpakowuje pary słownika jako argumenty nazwane.
opisz_pomiar(*lista, **slownik)

# Można też pominąć domyślny kanał i nie podawać odczytów.
opisz_pomiar("PR-206", jednostka="mV")
```

## Przekazywanie parametrów - slajd 29

* Dlaczego `append()` zmienia listę widoczną poza funkcją, a przypisanie nowej listy do parametru jej nie zastępuje?

```python
def dopisz_odczyt(odczyty):
    odczyty.append(1.4)

def przypisz_nowa_liste(odczyty):
    odczyty = [99.0]
    print("Nowa lista wewnątrz funkcji:", odczyty)

seria = [1.2, 1.3]
print("Przed wywołaniami:", seria)

dopisz_odczyt(seria)
print("Po dopisaniu odczytu:", seria)

przypisz_nowa_liste(seria)
print("Po przypisaniu nowej listy wewnątrz funkcji:", seria)
```

## Zwracane wartości - slajd 30

Zakładamy, że przekazana lista odczytów jest niepusta. Funkcja zwraca jeden obiekt: krotkę z trzema wynikami.

* W jakiej kolejności należy rozpakować krotkę? Czy nazwy zmiennych przy rozpakowaniu muszą odpowiadać nazwom użytym wewnątrz funkcji?


```python
def podsumuj_pomiary(odczyty):
    najmniejszy = min(odczyty)
    najwiekszy = max(odczyty)
    srednia = sum(odczyty) / len(odczyty)
    return najmniejszy, najwiekszy, srednia

seria = [10.0, 12.0, 11.0]
podsumowanie = podsumuj_pomiary(seria)
print("Zwrócona krotka:", podsumowanie)
print("Typ wyniku:", type(podsumowanie))

minimum, maksimum, srednia_serii = podsumowanie
print(f"Minimum: {minimum:.1f} mV")
print(f"Maksimum: {maksimum:.1f} mV")
print(f"Średnia: {srednia_serii:.1f} mV")
```
