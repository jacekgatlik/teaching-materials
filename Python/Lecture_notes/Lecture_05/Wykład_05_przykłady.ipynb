## Importowanie modułów i aliasy - slajd 4

Na slajdzie importujemy własny moduł `matematyka`. Tutaj te same trzy formy importu pokazujemy na gotowym module `math`, aby przykład działał bez dodatkowego pliku. Po `import math` odwołujemy się do funkcji przez nazwę modułu; po `from math import sqrt` możemy używać samego `sqrt`. Alias `m` jest inną nazwą tego samego modułu.

* Jak zmienia się zapis wywołania funkcji w każdej z trzech form?

```python
import math

wynik = math.sqrt(16)
print("Import modułu:", wynik)

from math import sqrt

wynik = sqrt(16)
print("Import wybranej funkcji:", wynik)

import math as m

wynik = m.sqrt(16)
print("Import z aliasem:", wynik)
```

## Funkcje specjalne i odległość punktów - slajd 19

`math.factorial()` oblicza silnię nieujemnej liczby całkowitej, a `math.gcd()` największy wspólny dzielnik liczb całkowitych. W funkcji `distance()` rozpakowujemy współrzędne dwóch punktów z krotek. `math.hypot()` oblicza długość odcinka na podstawie różnic współrzędnych.

```python
import math

print("5! =", math.factorial(5))
print("NWD liczb 48 i 180:", math.gcd(48, 180))

def distance(a, b):
    x1, y1 = a
    x2, y2 = b
    return math.hypot(x2 - x1, y2 - y1)

point_a = (0, 0)
point_b = (3, 4)

print("Odległość A–B:", distance(point_a, point_b))
print("Odległość B–A:", distance(point_b, point_a))
```

## Średnie i analiza wyników - slajd 23

Dla dodatnich danych obliczamy średnią harmoniczną i geometryczną, a następnie analizujemy przykładowe wyniki egzaminu. `statistics.stdev()` oblicza odchylenie standardowe próby. Wszystkie wyniki egzaminu występują po jednym razie: `mode()` rozstrzyga taki remis, zwracając pierwszą napotkaną wartość; nie oznacza to, że 88 występuje częściej od pozostałych wyników.

```python
import statistics

data = [40, 60, 80]
print("Średnia harmoniczna:", statistics.harmonic_mean(data))

data = [1, 3, 9, 27]
print("Średnia geometryczna:", statistics.geometric_mean(data))

scores = [88, 92, 79, 93, 85, 91, 76, 89, 95, 90]
mean_score = statistics.mean(scores)
median_score = statistics.median(scores)
mode_score = statistics.mode(scores)
stdev_score = statistics.stdev(scores)

print(f"Średnia: {mean_score}")
print(f"Mediana: {median_score}")
print(f"Moda: {mode_score}")
print(f"Odchylenie standardowe próby: {stdev_score}")
```

## Powtarzalna symulacja rzutów kostką - slajd 28

`random.randint(1, 6)` losuje liczbę całkowitą od 1 do 6 włącznie. Ponownie ustawiamy to samo ziarno przed drugą serią. W tym samym środowisku, przy tej samej kolejności wywołań generatora, uzyskujemy te same wyniki.

* Dlaczego druga seria powtarza pierwszą? Co się stanie, jeśli usuniemy drugie wywołanie `random.seed(42)`?

```python
import random

def roll_dice():
    return random.randint(1, 6)

random.seed(42)
pierwsza_seria = []
print("Pierwsza seria:")
for _ in range(6):
    wynik = roll_dice()
    pierwsza_seria.append(wynik)
    print(wynik)

#random.seed(42)
druga_seria = [roll_dice() for _ in range(10)]
print("Druga seria:", druga_seria)
print("Zgodność serii:", pierwsza_seria == druga_seria)
```

## Kilka serii i wykres funkcji bez NumPy - slajd 38

* Dlaczego do narysowania modelu używamy większej liczby punktów niż dla pomiarów? Co oznacza argument `'o'`?

```python
import math
import matplotlib.pyplot as plt

czas = [i / 10 for i in range(51)]
model = [100 * math.exp(-t / 2) for t in czas]
czas_pomiaru = [0, 1, 2, 3, 4, 5]
sygnal = [101, 59, 38, 21, 14, 8]

plt.figure(figsize=(7, 4))
plt.plot(czas, model, label='Model')
plt.plot(czas_pomiaru, sygnal, 'o', label='Pomiary')
plt.xlabel('Czas [s]')
plt.ylabel('Sygnał [j.u.]')
plt.legend()
plt.grid(alpha=0.9)
plt.tight_layout()
plt.show()
```

## Dwa wykresy na jednej figurze - slajd 43

* Co zmienia liczba przedziałów `bins`? 

```python
import random
import matplotlib.pyplot as plt

random.seed(42)
dane = [random.gauss(0, 1) for _ in range(1000)]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(9, 4))
ax1.hist(dane, bins=25, edgecolor='black')
ax1.set_title('Histogram')
ax1.set_xlabel('Wartość [j.u.]')
ax1.set_ylabel('Liczebność')
ax2.boxplot(dane)
ax2.set_title('Wykres pudełkowy')
ax2.set_ylabel('Wartość [j.u.]')
fig.tight_layout()
plt.show()
```

## Niepewności pomiarowe - slajd 44

* Jak zmieniłby się wykres po podaniu jednej liczby zamiast listy w `yerr`?

```python
import matplotlib.pyplot as plt

czas = [0, 1, 2, 3, 4, 5]
sygnal = [101, 59, 38, 21, 14, 8]
niepewnosc = [5, 4, 3, 2, 2, 1]

plt.figure(figsize=(7, 4))
plt.errorbar(czas, sygnal, yerr=niepewnosc,
             fmt='o', capsize=4, label='Pomiary')
plt.xlabel('Czas [s]')
plt.ylabel('Sygnał [j.u.]')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Zapis wykresu do pliku - slajd 46

* Który format zachowa ostre linie i tekst przy dużym powiększeniu? Co zmieni zwiększenie `dpi` dla PNG?

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(['A', 'B', 'C', 'D'], [5, 7, 3, 4])
ax.set_title('Porównanie grup')
ax.set_xlabel('Grupa')
ax.set_ylabel('Wartość [j.u.]')
fig.tight_layout()
fig.savefig('wykres.png', dpi=300, bbox_inches='tight')
fig.savefig('wykres.pdf', bbox_inches='tight')
plt.show()
plt.close(fig)
```
