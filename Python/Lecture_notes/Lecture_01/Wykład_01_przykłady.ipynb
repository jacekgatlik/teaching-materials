## Operatory arytmetyczne - slajd 33

```python
czas_s = 3671

godziny = czas_s // 3600
minuty = (czas_s % 3600) // 60
sekundy = czas_s % 60

print(f"Czas: {godziny:02d}:{minuty:02d}:{sekundy:02d}")
print(f"Czas w minutach: {czas_s / 60:.2f}")
print("Odtworzona liczba sekund:", godziny * 3600 + minuty * 60 + sekundy)
```

## Przykłady - slajd 40

```python
tekst_zliczen = input("Liczba zliczeń: ")
print("Typ przed konwersją:", type(tekst_zliczen))

zliczenia = int(tekst_zliczen)
czas_s = float(input("Czas pomiaru [s]: "))

if zliczenia < 0 or czas_s <= 0:
    print("Liczba zliczeń musi być nieujemna, a czas dodatni.")
else:
    czestosc = zliczenia / czas_s
    print(f"Średnia częstość zliczeń: {czestosc:.2f} 1/s")
```

## Przykłady - slajd 43

* Do której gałęzi trafi wynik równy dokładnie 105?

```python
sygnal = 105.0
dolna_granica = 95.0
gorna_granica = 105.0

if sygnal < dolna_granica:
    print("Sygnał za mały")
elif sygnal <= gorna_granica:
    print("Sygnał w przedziale")
else:
    print("Sygnał za duży")

w_przedziale = sygnal >= dolna_granica and sygnal <= gorna_granica
print("W przedziale:", w_przedziale)
```

## Pętla while - slajd 45

Obserwujemy aktywność wyłącznie po kolejnych pełnych okresach połowicznego zaniku. Startujemy od 800 jednostek umownych i szukamy pierwszego kroku z aktywnością nie większą niż 100. Nie wyznaczamy czasu ciągłego.

* Ile razy wykona się pętla? Czy wartość równa progowi kończy obliczenia?

```python
aktywnosc = 800.0
prog = 100.0
okresy = 0

print(f"Okres {okresy}: {aktywnosc:.1f}")
while aktywnosc > prog:
    aktywnosc = aktywnosc / 2
    okresy += 1
    print(f"Okres {okresy}: {aktywnosc:.1f}")

print(f"Potrzebne pełne okresy: {okresy}")
```

## Pętla for - slajd 48

Zbieramy trzy odczyty; do średniej wystarczy bieżąca suma i liczba pomiarów. `range(1, n + 1)` numeruje pomiary od 1 do n.

* Gdzie trzeba wyzerować sumę — przed pętlą czy w jej środku?

```python
n = 3
suma = 0.0

if n <= 0:
    print("Liczba pomiarów musi być dodatnia.")
else:
    for numer in range(1, n + 1):
        pomiar = float(input(f"Pomiar {numer}: "))
        suma += pomiar
        print(f"Suma po {numer} pomiarach: {suma:.2f}")

    srednia = suma / n
    print(f"Średnia: {srednia:.2f}")
```

## Podsumowanie

* Dlaczego `stop` sprawdzamy przed konwersją na liczbę?

```python
suma = 0.0
liczba = 0

while True:
    tekst = input("Nieujemny odczyt albo stop: ")
    if tekst == "stop":
        break

    odczyt = float(tekst)
    if odczyt < 0:
        print("Pomijam ujemny odczyt.")
        continue

    suma += odczyt
    liczba += 1
    print(f"Przyjęto odczyt numer {liczba}")

if liczba > 0:
    print(f"Średnia z {liczba} odczytów: {suma / liczba:.2f}")
else:
    print("Brak odczytów do obliczenia średniej.")
```
