## Indeksowanie i wycinki tekstu - slajd 6

## Formatowanie wyników w tekście - slajd 12

* Czy zapis `:.2f` zmienia wartość zmiennej, czy tylko sposób jej wyświetlenia?

```python
kod = "  pr-204  ".strip().upper()
tekst_wartosci = "12,345"
wartosc = float(tekst_wartosci.replace(",", "."))
jednostka = "mV"

print(f"Próbka {kod}: {wartosc:.2f} {jednostka}")
print("Wartość po formatowaniu:", wartosc)
```

## Dodawanie elementów do listy - slajd 18

* Co doda `append()`, gdy argumentem jest lista? Ile elementów będzie miała każda z list wynikowych?


```python
przez_append = [10, 12]
przez_extend = [10, 12]
nowe_odczyty = [11, 13]

przez_append.append(nowe_odczyty)
przez_extend.extend(nowe_odczyty)

print("append:", przez_append, "— długość:", len(przez_append))
print("extend:", przez_extend, "— długość:", len(przez_extend))
```

## Wyrażenia listowe i numerowanie - slajd 22

* Czy `enumerate(..., start=1)` numeruje zaakceptowane wyniki, czy zachowuje ich pozycje na liście wejściowej?

```python
odczyty_mV = [1200, -5, 980, 1500]
odczyty_V = [wartosc / 1000 for wartosc in odczyty_mV if wartosc >= 0]

print("Dane wejściowe [mV]:", odczyty_mV)
for numer, wartosc in enumerate(odczyty_V, start=1):
    print(f"Wynik {numer}: {wartosc:.3f} V")
```

## Kopia płytka i głęboka - slajd 23

* Które zmienne wskazują tę samą listę? Które listy współdzielą zagnieżdżone elementy?


```python
from copy import deepcopy

pomiary = [[1.0, 1.1], [2.0, 2.1]]
alias = pomiary
plytka = pomiary.copy()
gleboka = deepcopy(pomiary)

alias.append([3.0, 3.1])
print("Oryginał:", pomiary)
print("Kopia płytka:", plytka)
print("Alias to oryginał:", alias is pomiary)
print("Kopia płytka to oryginał:", plytka is pomiary)
```

```python
plytka[0][0] = 99.0

print("Oryginał po zmianie wewnętrznej listy:", pomiary)
print("Kopia płytka:", plytka)
print("Kopia głęboka:", gleboka)
```

## Krotki z obiektami mutowalnymi - slajd 33

* Czy rozpakowanie krotki kopiuje listę? Dlaczego można dopisać odczyt, choć sama krotka jest niemutowalna?


```python
rekord = ("PR-204", [1.2, 1.3])
kod, odczyty = rekord
odczyty.append(1.4)

print("Kod próbki:", kod)
print("Krotka po dopisaniu odczytu:", rekord)
print("To ta sama lista:", odczyty is rekord[1])

rekord[1][0] = 2.1
print("Krotka po dopisaniu odczytu:", rekord)
#Ale - TypeError:
rekord[1] = [3.1, 4.5]
```

## Dostęp do wartości słownika - slajd 38

* Czy `get()` z wartością domyślną dopisuje brakujący klucz?


```python
pomiar = {"probka": "PR-204", "wartosc": 12.5}

print("Wartość:", pomiar["wartosc"])
print("Jednostka:", pomiar.get("jednostka", "brak danych"))
print("Czy jest klucz jednostka?", "jednostka" in pomiar)
print("Klucze po get():", list(pomiar))
```

```python
pomiar["jednostka"] = "mV"
pomiar["wartosc"] = 12.6  # Aktualizacja nie zmienia położenia klucza.

for klucz, wartosc in pomiar.items():
    print(f"{klucz}: {wartosc}")
```

## Suma, część wspólna i różnica zbiorów - slajd 48

Porównujemy oznaczenia próbek z dwóch serii. `sorted()` służy tu tylko do czytelnego, powtarzalnego wypisywania wyników.

* Które próbki występują w obu seriach, a które tylko w jednej? 


```python
seria_A = set(["PR-01", "PR-02", "PR-02", "PR-03"])
seria_B = {"PR-02", "PR-03", "PR-04"}

print("Unikalne w A:", sorted(seria_A))
print("Wszystkie:", sorted(seria_A | seria_B))
print("W obu seriach:", sorted(seria_A & seria_B))
print("Tylko w A:", sorted(seria_A - seria_B))
print("Tylko w B:", sorted(seria_B - seria_A))
print("Dokładnie w jednej serii:", sorted(seria_A ^ seria_B))
```

## Wyszukiwanie i zamiana tekstu - slajd 61

* Co zwróci `search()`, jeśli nie znajdzie oznaczenia próbki? Dlaczego przed `group()` sprawdzamy wynik wyszukiwania?


```python
import re

for tekst in ("Próbka PR-204: odczyt 12.5 mV.", "Brak oznaczenia próbki."):
    dopasowanie = re.search(r"PR-\d+", tekst)
    if dopasowanie is not None:
        print("Oznaczenie:", dopasowanie.group())
        print("Zakres indeksów:", dopasowanie.span())
    else:
        print("Nie znaleziono oznaczenia.")
```

```python
tekst = "Próbka PR-204: odczyt 12.5 mV."
nowy_tekst = re.sub(r"PR-\d+", "PR-999", tekst)

print("Po zamianie:", nowy_tekst)
print("Oryginał:", tekst)
```
