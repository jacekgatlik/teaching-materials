## Zadanie 4.1 — Przybliżenie sinusa szeregiem Taylora (4 punkty)

Napisz program obliczający przybliżoną wartość sinusa kąta $x$
podanego w radianach, korzystając z pierwszych $n$ wyrazów szeregu Taylora.
Parametr $n$ oznacza **liczbę sumowanych wyrazów**, a nie indeks
ostatniego wyrazu.

### Wymagania

1. Pobierz od użytkownika skończoną liczbę rzeczywistą $x$
   oraz dodatnią liczbę całkowitą $n$. Dla niepoprawnych danych
   wyświetl komunikat i poproś o ponowne podanie wartości.
2. Napisz funkcję `taylor_sin(x, n)`, która oblicza sumę:

   $$
   S_n(x) = \sum_{k=0}^{n-1}\frac{(-1)^k x^{2k+1}}{(2k+1)!}.
   $$

   Silnię obliczaj w odrębnej funkcji `silnia(m)`, napisanej
   z użyciem pętli lub rekurencji. Przyjmij całkowite $m \geq 0$
   i pamiętaj, że $0! = 1$. Nie używaj `math.factorial()`.
3. Z funkcji `taylor_sin(x, n)` zwróć krotkę zawierającą
   przybliżenie oraz błąd bezwzględny $|S_n(x) - \sin(x)|$,
   obliczony względem `math.sin(x)`. Wyświetl obie wartości.

### Punktacja

| Kryterium | Punkty |
|---|---:|
| Pobranie poprawnych wartości $x$ i $n$ (1) | 1.0 |
| Obliczenie sumy w `taylor_sin()` z użyciem własnej funkcji silni (2) | 2.0 |
| Zwrócenie krotki z przybliżeniem i błędem oraz wyświetlenie wyników (3) | 1.0 |
| **Łącznie** | **4.0** |

### Wskazówka

Funkcja `math.sin()` z modułu `math` przyjmuje kąt w radianach.
Wartość bezwzględną oblicza `abs()`. Do sprawdzenia programu
można użyć np. `x = 0.5` i `n = 5`.

````python
import math

def silnia(m):
    wynik = 1
    for i in range(2, m + 1):
        wynik *= i
    return wynik

def taylor_sin(x, n):
    suma = 0.0
    for k in range(n):
        suma += (-1) ** k * x ** (2 * k + 1) / silnia(2 * k + 1)
    blad = abs(suma - math.sin(x))
    return suma, blad

while True:
    try:
        x = float(input("Podaj x w radianach: "))
        n = int(input("Podaj liczbę wyrazów n: "))
        if math.isfinite(x) and n > 0:
            break
        print("x musi być liczbą skończoną, a n musi być dodatnie.")
    except ValueError:
        print("Podaj liczbę rzeczywistą x i liczbę całkowitą n.")

przyblizenie, blad = taylor_sin(x, n)
print("Przybliżenie sinusa:", przyblizenie)
print("Błąd bezwzględny:", blad)
````

## Zadanie 4.2 — Przybliżenie liczby pi metodą Monte Carlo (4 punkty)

Napisz program, który szacuje wartość liczby $\pi$ przez losowanie
punktów w kwadracie $[-1,1] \times [-1,1]$.
Koło o środku $(0,0)$ i promieniu 1 ma pole $\pi$, a kwadrat ma pole 4.

### Wymagania

1. Pobierz od użytkownika dodatnią liczbę całkowitą $n$,
   oznaczającą liczbę losowanych punktów. Dla niepoprawnych danych
   wyświetl komunikat i poproś o ponowne podanie wartości.
2. Napisz funkcję `oblicz_pi(n)`, która losuje $n$ punktów.
   Dla każdego punktu wylosuj niezależnie współrzędne $x$ i $y$
   z rozkładu jednostajnego na przedziale $[-1,1]$.
   Zlicz punkty spełniające warunek $x^2 + y^2 \leq 1$.
3. W funkcji `oblicz_pi(n)` oblicz przybliżenie:

   $$
   \pi_{\mathrm{MC}} = 4\frac{\text{liczba punktów w kole}}{n}.
   $$

   Zwróć krotkę zawierającą przybliżenie oraz błąd bezwzględny
   względem `math.pi`.
4. Wyświetl liczbę wylosowanych punktów, przybliżenie $\pi$
   oraz błąd bezwzględny.

### Punktacja

| Kryterium | Punkty |
|---|---:|
| Pobranie dodatniej, całkowitej liczby punktów (1) | 1.0 |
| Losowanie punktów i zliczanie punktów w kole (2) | 1.0 |
| Obliczenie i zwrócenie przybliżenia oraz błędu (3) | 1.0 |
| Czytelne wyświetlenie wyników (4) | 1.0 |
| **Łącznie** | **4.0** |

### Wskazówka

Liczbę z rozkładu jednostajnego można wylosować funkcją
`random.uniform(-1, 1)` z modułu `random`.
Wyniki kolejnych uruchomień mogą się różnić. Większe $n$
zwykle poprawia dokładność oszacowania, ale nie gwarantuje
mniejszego błędu w każdej pojedynczej próbie.

````python
import math
import random

def oblicz_pi(n):
    punkty_w_kole = 0
    for i in range(n):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if x ** 2 + y ** 2 <= 1:
            punkty_w_kole += 1
    przyblizenie = 4 * punkty_w_kole / n
    blad = abs(przyblizenie - math.pi)
    return przyblizenie, blad

while True:
    try:
        n = int(input("Podaj liczbę punktów n: "))
        if n > 0:
            break
        print("Liczba punktów musi być dodatnia.")
    except ValueError:
        print("Podaj dodatnią liczbę całkowitą.")

przyblizenie, blad = oblicz_pi(n)
print("Liczba punktów:", n)
print("Przybliżenie pi:", przyblizenie)
print("Błąd bezwzględny:", blad)
````

   pomiary = [("D1", 2.1), ("D2", 1.7), ("D3", 2.0),

### Punktacja

| Wyświetlenie oznaczeń, dawek i odchyleń z jednostkami (3) | 1.0 |
| **Łącznie** | **4.0** |

### Wskazówka

W krotce `pomiar` dawka znajduje się pod indeksem `1`.
Funkcja przekazana jako `key` wyznacza wartość używaną
do porównywania elementów. Wartość bezwzględną oblicza `abs()`.

````python
def uporzadkuj_pomiary(pomiary, dawka_odniesienia):
    return sorted(pomiary, key=lambda pomiar: abs(pomiar[1] - dawka_odniesienia))

pomiary = [("D1", 2.1), ("D2", 1.7), ("D3", 2.0),
           ("D4", 2.4), ("D5", 1.8)]
dawka_odniesienia = 2.0

uporzadkowane = uporzadkuj_pomiary(pomiary, dawka_odniesienia)

for nazwa, dawka in uporzadkowane:
    odchylenie = abs(dawka - dawka_odniesienia)
    print(f"{nazwa}: dawka = {dawka:.2f} mGy, odchylenie = {odchylenie:.2f} mGy")
````

### Wskazówka

Przed uruchomieniem programu przygotuj plik `halas.txt`
w bieżącym katalogu notebooka, np. z poniższą zawartością.

```text
biblioteka 25
korytarz 60
pracownia 75
czytelnia 33

```

Do otwarcia pliku użyj `with open(..., encoding="utf-8")`.

````python
grupy = {"niska": 0, "srednia": 0, "wysoka": 0}

try:
    with open("halas.txt", "r", encoding="utf-8") as plik:
        for wiersz in plik:
            if wiersz.strip():
                miejsce, tekst = wiersz.split()
                wartosc = int(tekst)
                if wartosc <= 33:
                    grupy["niska"] += 1
                elif wartosc <= 66:
                    grupy["srednia"] += 1
                else:
                    grupy["wysoka"] += 1
except FileNotFoundError:
    print("Nie znaleziono pliku halas.txt.")
else:
    print(grupy)
````
