## Zadanie 3.1 — Analiza pomiarów tętna (3 punkty)

Napisz program, który zarządza listą pomiarów tętna pacjentów.
W tym zadaniu dla pomiarów spoczynkowych przyjmij progi: powyżej
100 uderzeń/min — tachykardia, poniżej 60 uderzeń/min — bradykardia.

### Wymagania

1. Użyj listy do przechowywania pomiarów tętna, np.:

   ```python
   pomiary = [72, 85, 90, 65, 78]
   ```

2. Oblicz i wyświetl średnie, maksymalne i minimalne tętno.
3. Zidentyfikuj i wyświetl osobno pomiary większe niż 100 uderzeń/min
   oraz mniejsze niż 60 uderzeń/min. Wartości równe 60 i 100 nie należą
   do tych grup.
4. Pozwól użytkownikowi dopisywać pomiary za pomocą `input()`,
   aż wprowadzi `-1`. Nie dopisuj `-1` do listy. Przyjmuj dodatnie
   liczby całkowite; po błędnym wpisie wyświetl komunikat i poproś
   o kolejny pomiar.
5. Posortuj pełną listę malejąco i wyświetl wynik.

Analizę z punktów 2–3 wykonaj po zakończeniu dopisywania pomiarów,
aby wyniki uwzględniały także nowe dane.

### Punktacja

| Kryterium | Punkty |
|---|---:|
| Przechowywanie pomiarów w liście oraz obliczenie średniej, minimum i maksimum (1–2) | 1.0 |
| Wskazanie pomiarów powyżej 100 i poniżej 60 uderzeń/min (3) | 1.0 |
| Dopisywanie pomiarów do `-1` i sortowanie malejąco (4–5) | 1.0 |
| **Łącznie** | **3.0** |

### Wskazówka

Do obliczenia średniej można wykorzystać moduł `statistics`:

```python
import statistics

srednia = statistics.mean(pomiary)
```

Minimum i maksimum wyznaczysz funkcjami `min()` i `max()`.


````python
import statistics

pomiary = [72, 85, 90, 65, 78]

while True:
    try:
        tetno = int(input("Podaj tętno lub -1, aby zakończyć: "))
        if tetno == -1:
            break
        if tetno > 0:
            pomiary.append(tetno)
        else:
            print("Tętno musi być dodatnią liczbą całkowitą.")
    except ValueError:
        print("Podaj liczbę całkowitą.")

tachykardia = []
bradykardia = []

for tetno in pomiary:
    if tetno > 100:
        tachykardia.append(tetno)
    if tetno < 60:
        bradykardia.append(tetno)

print("Pomiary:", pomiary)
print("Średnie tętno:", statistics.mean(pomiary), "uderzeń/min")
print("Maksymalne tętno:", max(pomiary), "uderzeń/min")
print("Minimalne tętno:", min(pomiary), "uderzeń/min")
print("Pomiary powyżej 100 uderzeń/min:", tachykardia)
print("Pomiary poniżej 60 uderzeń/min:", bradykardia)

pomiary.sort(reverse=True)
print("Pomiary malejąco:", pomiary)
````

## Zadanie 3.2 — Analiza pomiarów prędkości (4 punkty)

Napisz program, który analizuje pomiary prędkości cząstek
w eksperymencie fizycznym.

### Wymagania

1. Pobierz od użytkownika co najmniej dwa pomiary prędkości w m/s,
   oddzielone przecinkami, np. `10.0, 10.1, 9.9, 10.0, 10.2, 10.1, 9.8, 30.0`.
   Ułamki zapisuj z kropką. Dla błędnego zapisu lub zbyt małej liczby

2. Napisz funkcję `analizuj_pomiary(pomiary)`, która zwraca krotkę

   - średnią prędkość,

3. Napisz funkcję `filtruj_pomiary(pomiary)`, która zwraca nową listę,
   pomijając pomiary oddalone od średniej o więcej niż dwa odchylenia
   standardowe. Oblicz średnią i odchylenie dla listy wejściowej

   Nie zmieniaj listy wejściowej.
4. Utwórz zbiór wartości prędkości, które występują więcej niż raz
   w oryginalnych pomiarach.

### Punktacja

| Kryterium | Punkty |
|---|---:|
| Pobranie pomiarów i funkcja zwracająca krotkę statystyk (1–2) | 1.0 |
| Jednorazowe odfiltrowanie wartości odstających (3) | 1.0 |
| Utworzenie zbioru powtarzających się wartości (4) | 1.0 |
| Czytelne wyświetlenie wszystkich wyników (5) | 1.0 |
| **Łącznie** | **4.0** |

### Wskazówka

Aby obliczyć średnią, medianę oraz odchylenie standardowe, można
skorzystać z modułu `statistics`:

```python
import statistics

srednia = statistics.mean(pomiary)
mediana = statistics.median(pomiary)
odchylenie = statistics.stdev(pomiary)
```

Funkcja `statistics.stdev()` oblicza odchylenie standardowe z próby
i wymaga co najmniej dwóch pomiarów. Wartość bezwzględną oblicza `abs()`.


````python
import statistics

def analizuj_pomiary(pomiary):
    srednia = statistics.mean(pomiary)
    mediana = statistics.median(pomiary)
    odchylenie = statistics.stdev(pomiary)
    return srednia, mediana, odchylenie

def filtruj_pomiary(pomiary):
    srednia, mediana, odchylenie = analizuj_pomiary(pomiary)
    oczyszczone = []
    for pomiar in pomiary:
        if abs(pomiar - srednia) <= 2 * odchylenie:
            oczyszczone.append(pomiar)
    return oczyszczone

while True:
    try:
        tekst = input("Podaj co najmniej dwie prędkości w m/s, oddzielone przecinkami: ")
        pomiary = []
        for element in tekst.split(","):
            pomiary.append(float(element))
        if len(pomiary) >= 2:
            break
        print("Potrzebne są co najmniej dwa pomiary.")
    except ValueError:
        print("Podaj liczby oddzielone przecinkami. Użyj kropki dziesiętnej.")

srednia, mediana, odchylenie = analizuj_pomiary(pomiary)
oczyszczone = filtruj_pomiary(pomiary)
powtorzenia = set()

for pomiar in pomiary:
    if pomiary.count(pomiar) > 1:
        powtorzenia.add(pomiar)

print("Pomiary:", pomiary)
print("Średnia prędkość:", srednia, "m/s")
print("Mediana prędkości:", mediana, "m/s")
print("Odchylenie standardowe z próby:", odchylenie, "m/s")
print("Powtarzające się prędkości:", powtorzenia)
print("Oczyszczone pomiary:", oczyszczone)
````

## Zadanie 3.3 — Analiza danych z detektora promieniowania (4 punkty)

Napisz program, który analizuje liczbę zliczeń z detektora promieniowania.
Każdy pomiar obejmuje osobny przedział czasu o długości 0.1 s.

**Założenie modelu:** detektor rejestruje dokładnie jedno zliczenie
na każdy rozpad; pomijamy tło i straty detekcji. W tym uproszczeniu
aktywność można oszacować jako $A \approx N / \Delta t$, gdzie
$N$ oznacza liczbę zliczeń, a $\Delta t = 0.1\,\mathrm{s}$.
Jednostką aktywności jest Bq (jeden rozpad na sekundę). Bez tych
założeń iloraz $N/\Delta t$ opisuje częstość zliczeń, a nie bezpośrednio
aktywność źródła.

### Wymagania

1. Pobierz od użytkownika niepustą listę nieujemnych, całkowitych
   liczb zliczeń oddzielonych przecinkami, np. `8, 12, 10, 15, 9`.
   Dla błędnych danych wyświetl komunikat zamiast wykonywać obliczenia.
2. Napisz funkcję `przeksztalc_dane(dane)`, która zwraca nową listę
   oszacowań aktywności w Bq, dzieląc każdą liczbę zliczeń przez 0.1 s.
3. Napisz funkcję `filtruj_dane(dane, prog=100.0)`, która przyjmuje
   listę aktywności w Bq i zwraca nową listę wartości nie mniejszych
   niż próg. Użyj przyjętego w zadaniu progu 100 Bq. Nie zmieniaj
   listy wejściowej.
4. Oblicz średnią aktywność i odchylenie standardowe z próby
   dla przefiltrowanych danych. Jeśli lista jest pusta, wyświetl
   komunikat o braku danych. Jeśli pozostał jeden pomiar, oblicz
   średnią i poinformuj, że do odchylenia standardowego potrzeba
   co najmniej dwóch pomiarów.
5. Wyświetl wszystkie oszacowania aktywności, listę po filtracji
   i otrzymane statystyki, podając jednostkę Bq.

### Punktacja

| Kryterium | Punkty |
|---|---:|
| Pobranie zliczeń i przekształcenie ich na aktywności w przyjętym modelu (1–2) | 1.0 |
| Odfiltrowanie wartości poniżej przyjętego progu (3) | 1.0 |
| Obliczenie statystyk z uwzględnieniem liczby pozostałych pomiarów (4) | 1.0 |
| Czytelne wyświetlenie wyników z jednostkami (5) | 1.0 |
| **Łącznie** | **4.0** |

### Wskazówka

Do obliczenia średniej oraz odchylenia standardowego można użyć
modułu `statistics`:

```python
import statistics

srednia = statistics.mean(przefiltrowane)
odchylenie = statistics.stdev(przefiltrowane)
```

Średnia wymaga co najmniej jednego pomiaru. Funkcja `statistics.stdev()`
oblicza odchylenie standardowe z próby i wymaga co najmniej dwóch pomiarów.


````python
import statistics

def przeksztalc_dane(dane):
    aktywnosci = []
    for zliczenia in dane:
        aktywnosci.append(zliczenia / 0.1)
    return aktywnosci

def filtruj_dane(dane, prog=100.0):
    przefiltrowane = []
    for aktywnosc in dane:
        if aktywnosc >= prog:
            przefiltrowane.append(aktywnosc)
    return przefiltrowane

while True:
    try:
        tekst = input("Podaj liczby zliczeń w oknach 0.1 s, oddzielone przecinkami: ")
        dane = []
        poprawne = True
        for element in tekst.split(","):
            zliczenia = int(element)
            if zliczenia < 0:
                poprawne = False
            dane.append(zliczenia)
        if poprawne:
            break
        print("Liczby zliczeń nie mogą być ujemne.")
    except ValueError:
        print("Podaj co najmniej jedną nieujemną liczbę całkowitą. Oddziel liczby przecinkami.")

aktywnosci = przeksztalc_dane(dane)
przefiltrowane = filtruj_dane(aktywnosci)

print("Aktywności w modelu idealnym (Bq):", aktywnosci)
print("Przyjęty próg: 100 Bq")
print("Aktywności po filtracji (Bq):", przefiltrowane)

if len(przefiltrowane) == 0:
    print("Brak pomiarów po filtracji. Nie można obliczyć średniej ani odchylenia standardowego.")
else:
    print("Średnia aktywność:", statistics.mean(przefiltrowane), "Bq")
    if len(przefiltrowane) >= 2:
        print("Odchylenie standardowe z próby:", statistics.stdev(przefiltrowane), "Bq")
    else:
        print("Do obliczenia odchylenia standardowego z próby potrzebne są co najmniej dwa pomiary.")
````

## Zadanie 3.4 — Całkowanie metodą prostokątów (4 punkty)

Napisz program, który oblicza przybliżoną wartość całki funkcji
$f(x) = x^2 \sin(x)$ metodą prostokątów środkowych.

### Wymagania

1. Napisz funkcję `funkcja(x)`, która zwraca wartość $f(x) = x^2 \sin(x)$.
   Argument funkcji sinus podawany jest w radianach.
2. Napisz funkcję `calka_prostokaty(a, b, n)`, która dzieli przedział
   $[a,b]$ na $n$ równych części i zwraca przybliżoną wartość całki.
   Wysokość każdego prostokąta wyznaczaj w środku podprzedziału:

   $$
   h = \frac{b-a}{n}, \qquad x_i = a + \left(i + \frac{1}{2}\right)h,
   \qquad i = 0, 1, \ldots, n-1.
   $$

   Przybliżenie całki to suma $f(x_i)h$ dla wszystkich podprzedziałów.
3. Pobierz od użytkownika:
   - $a$ — początek przedziału całkowania,
   - $b$ — koniec przedziału całkowania,
   - $n$ — dodatnią, całkowitą liczbę podprzedziałów.

   Przyjmij $a < b$. Dla niepoprawnych danych wyświetl komunikat
   zamiast wykonywać obliczenia.
4. Oblicz i wyświetl przybliżoną wartość całki.

### Punktacja

| Kryterium | Punkty |
|---|---:|
| Zdefiniowanie funkcji $f(x)$ (1) | 1.0 |
| Implementacja metody prostokątów środkowych (2) | 1.0 |
| Pobranie poprawnych wartości $a$, $b$, $n$ (3) | 1.0 |
| Obliczenie i wyświetlenie wartości całki (4) | 1.0 |
| **Łącznie** | **4.0** |

### Wskazówka

Funkcja `math.sin()` z modułu `math` oblicza sinus kąta w radianach:

```python
import math

wartosc = math.sin(math.pi / 2)
```


````python
import math

def funkcja(x):
    return x ** 2 * math.sin(x)

def calka_prostokaty(a, b, n):
    h = (b - a) / n
    suma = 0
    for i in range(n):
        x = a + (i + 0.5) * h
        suma += funkcja(x)
    return suma * h

while True:
    try:
        a = float(input("Podaj początek przedziału a: "))
        b = float(input("Podaj koniec przedziału b: "))
        n = int(input("Podaj liczbę podprzedziałów n: "))
        if a < b and n > 0:
            break
        print("Początek przedziału musi być mniejszy od końca, a n musi być dodatnie.")
    except ValueError:
        print("Podaj liczby a i b oraz całkowite n. Użyj kropki dziesiętnej.")

wynik = calka_prostokaty(a, b, n)
print("Przybliżona wartość całki:", wynik)
````
