## Zadanie 5.1 — Kontrola położenia fantomu (3 punkty)

Plik `pozycje_fantomu.txt` zawiera przesunięcia fantomu względem ustalonego położenia. Na potrzeby ćwiczenia przyjmij tolerancję 1,0 mm w każdej z dwóch osi. Przygotuj raport zawierający pomiary poza tolerancją.

### Wymagania

1. Napisz funkcję `zapisz_przekroczenia(plik_wejsciowy, plik_wyjsciowy, tolerancja)`. Wczytaj w niej plik `pozycje_fantomu.txt`. Każdy niepusty wiersz zawiera identyfikator pomiaru oraz przesunięcia $x$ i $y$ w mm, oddzielone białymi znakami. Pomiń puste wiersze i przyjmij, że pozostałe dane są poprawne.
2. Zapisz w pliku `przekroczenia.txt` pomiary, dla których $|x|>\text{tolerancja}$ lub $|y|>\text{tolerancja}$. Zachowaj kolejność wierszy, identyfikatory i oba przesunięcia ze znakami. Wartość bezwzględna równa tolerancji mieści się w granicy dla danej osi.
3. Zwróć z funkcji liczbę zapisanych pomiarów. Wywołaj funkcję z tolerancją 1,0 mm i wyświetl wynik.

### Punktacja

| Kryterium | Punkty |
|---|---:|
| Odczyt i rozdzielenie danych w funkcji (1) | 1.0 |
| Zastosowanie warunku dla obu osi i zapis raportu (2) | 1.0 |
| Zwrócenie i wyświetlenie liczby pomiarów poza tolerancją (3) | 1.0 |
| **Łącznie** | **3.0** |

````python
def zapisz_przekroczenia(plik_wejsciowy, plik_wyjsciowy, tolerancja):
    liczba_przekroczen = 0

    with open(plik_wejsciowy, encoding="utf-8") as wejscie:
        with open(plik_wyjsciowy, "w", encoding="utf-8") as wyjscie:
            for wiersz in wejscie:
                if not wiersz.strip():
                    continue

                identyfikator, zapis_x, zapis_y = wiersz.split()
                x = float(zapis_x)
                y = float(zapis_y)

                if abs(x) > tolerancja or abs(y) > tolerancja:
                    wyjscie.write(f"{identyfikator} {zapis_x} {zapis_y}\n")
                    liczba_przekroczen += 1

    return liczba_przekroczen

wynik = zapisz_przekroczenia("pozycje_fantomu.txt", "przekroczenia.txt", 1.0)
print(f"Liczba pomiarów poza tolerancją: {wynik}")
````

## Zadanie 5.2 — Losowanie stanowisk do kontroli (4 punkty)

W pracowni należy wylosować trzy stanowiska do kontroli. Stanowiska D03 i D07 są chwilowo wyłączone. Jedno z trzech wylosowanych stanowisk przejdzie dodatkowy pomiar.

### Wymagania

1. Przyjmij poniższe dane. Napisz funkcję `wylosuj_kontrole(stanowiska, wylaczone, n)`, która tworzy listę dostępnych stanowisk w kolejności z listy wejściowej.

   ```python
   stanowiska = ["D01", "D02", "D03", "D04", "D05", "D06", "D07", "D08"]
   wylaczone = ["D03", "D07"]
   ```

2. W funkcji wylosuj `n` stanowisk bez zwracania, korzystając z `random.sample()`.
3. Wybierz za pomocą `random.choice()` jedno z wylosowanych stanowisk do dodatkowego pomiaru. Zwróć krotkę zawierającą listę wylosowanych stanowisk i identyfikator wybranego stanowiska.
4. Przed wywołaniem funkcji ustaw `random.seed(42)`. Wywołaj funkcję dla `n=3` i wyświetl oba wyniki.

### Punktacja

| Kryterium | Punkty |
|---|---:|
| Funkcja i poprawna lista dostępnych stanowisk (1) | 1.0 |
| Losowanie bez zwracania (2) | 1.0 |
| Wybór do dodatkowego pomiaru i zwrócenie obu wyników (3) | 1.0 |
| Ustawienie ziarna, wywołanie funkcji i wyświetlenie wyników (4) | 1.0 |
| **Łącznie** | **4.0** |

````python
import random

def wylosuj_kontrole(stanowiska, wylaczone, n):
    dostepne = []
    for stanowisko in stanowiska:
        if stanowisko not in wylaczone:
            dostepne.append(stanowisko)

    wylosowane = random.sample(dostepne, n)
    dodatkowy_pomiar = random.choice(wylosowane)
    return wylosowane, dodatkowy_pomiar

stanowiska = ["D01", "D02", "D03", "D04", "D05", "D06", "D07", "D08"]
wylaczone = ["D03", "D07"]

random.seed(42)
wylosowane, dodatkowy_pomiar = wylosuj_kontrole(stanowiska, wylaczone, 3)
print("Stanowiska do kontroli:", ", ".join(wylosowane))
print("Stanowisko do dodatkowego pomiaru:", dodatkowy_pomiar)
````

## Zadanie 5.3 — Chłodzenie fantomu (4 punkty)

Plik `chlodzenie_fantomu.txt` zawiera symulowane odczyty dwóch czujników temperatury. Jeden mierzy temperaturę w środku fantomu, drugi na jego powierzchni. Przedstaw obie serie na wspólnym wykresie.

### Wymagania

1. Wczytaj plik, pomijając nagłówek. Zapisz w trzech listach czasy w minutach, temperatury w środku oraz temperatury na powierzchni w °C. Kolumny są oddzielone białymi znakami.
2. Narysuj obie temperatury w funkcji czasu na wspólnych osiach. Użyj linii ze znacznikami pomiarów i odróżnij serie stylem linii lub znacznikiem.
3. Dodaj tytuł, opisy osi z jednostkami oraz legendę wskazującą położenie czujników.
4. Zapisz wykres jako `chlodzenie_fantomu.png` i wyświetl go.

### Punktacja

| Kryterium | Punkty |
|---|---:|
| Odczyt trzech kolumn i konwersja na liczby (1) | 1.0 |
| Dwie poprawne i rozróżnialne serie na wspólnych osiach (2) | 1.0 |
| Tytuł, opisy osi z jednostkami i legenda (3) | 1.0 |
| Zapis do PNG i wyświetlenie wykresu (4) | 1.0 |
| **Łącznie** | **4.0** |

````python
import matplotlib.pyplot as plt

czasy = []
temperatury_srodek = []
temperatury_powierzchnia = []

with open("chlodzenie_fantomu.txt", encoding="utf-8") as plik:
    next(plik)
    for wiersz in plik:
        czas, srodek, powierzchnia = wiersz.split()
        czasy.append(float(czas))
        temperatury_srodek.append(float(srodek))
        temperatury_powierzchnia.append(float(powierzchnia))

plt.figure()
plt.plot(czasy, temperatury_srodek, marker="o", label="Środek")
plt.plot(czasy, temperatury_powierzchnia, marker="s", label="Powierzchnia")
plt.title("Chłodzenie fantomu")
plt.xlabel("Czas [min]")
plt.ylabel("Temperatura [°C]")
plt.legend()
plt.tight_layout()
plt.savefig("chlodzenie_fantomu.png")
plt.show()
````

## Zadanie 5.4 — Porównanie profili czułości detektora (4 punkty)

W uproszczonym modelu odpowiedź detektora $R$, wyrażona w procentach
wartości maksymalnej, zależy od przesunięcia $x$ względem środka.
Parametr $w$ określa szerokość profilu.

$$
R(x,w)=100\exp\!\left[-(x/w)^2\right].
$$

### Wymagania

1. Napisz funkcję `odpowiedz(x, w)`, która oblicza i zwraca $R(x,w)$.
2. Przygotuj listę całkowitych przesunięć od −20 do 20 mm włącznie.
   Oblicz dwie listy odpowiedzi dla $w=4$ mm i $w=8$ mm.
3. Narysuj obie krzywe na wspólnych osiach. Dodaj tytuł, opisy osi
   z jednostkami i legendę wskazującą wartość $w$ dla każdej krzywej.
4. Na podstawie wykresu napisz jedno zdanie, dla którego $w$ odpowiedź
   jest większa przy przesunięciu o 5 mm od środka.

### Punktacja

| Kryterium | Punkty |
|---|---:|
| Funkcja obliczająca i zwracająca odpowiedź (1) | 1.0 |
| Listy przesunięć i odpowiedzi dla obu wartości $w$ (2) | 1.0 |
| Dwie krzywe na wspólnych osiach z opisami i legendą (3) | 1.0 |
| Poprawne odczytanie zależności z wykresu (4) | 1.0 |
| **Łącznie** | **4.0** |

````python
import math
import matplotlib.pyplot as plt

def odpowiedz(x, w):
    return 100 * math.exp(-(x / w) ** 2)

przesuniecia = list(range(-20, 21))
odpowiedzi_4 = []
odpowiedzi_8 = []

for x in przesuniecia:
    odpowiedzi_4.append(odpowiedz(x, 4))
    odpowiedzi_8.append(odpowiedz(x, 8))

plt.figure()
plt.plot(przesuniecia, odpowiedzi_4, label="w = 4 mm")
plt.plot(przesuniecia, odpowiedzi_8, linestyle="--", label="w = 8 mm")
plt.title("Profile czułości detektora")
plt.xlabel("Przesunięcie [mm]")
plt.ylabel("Odpowiedź [%]")
plt.legend()
plt.tight_layout()
plt.show()

print("Przy przesunięciu o 5 mm od środka odpowiedź jest większa dla w = 8 mm.")
````
