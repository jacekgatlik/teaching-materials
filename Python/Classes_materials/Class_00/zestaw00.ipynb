### Skąd ta różnica?

Python przechowuje te liczby jako `float`: liczby zmiennoprzecinkowe o ograniczonej precyzji. W typowym środowisku używanym na zajęciach jest to format binary64, dający około 15–17 cyfr znaczących. Nie oznacza to stałej liczby miejsc po przecinku.

Tak jak `1/3` nie ma skończonego rozwinięcia dziesiętnego, tak `0.1` nie ma skończonego rozwinięcia dwójkowego. Komputer przechowuje jego przybliżenie. Natomiast `0.5` i `0.25` można zapisać w tym formacie dokładnie.

Przybliżony zapis nie oznacza, że każde porównanie zakończy się wynikiem `False`: różne obliczenia mogą prowadzić do tej samej reprezentowanej wartości.

**Eksperyment:** w komórce poniżej zmień wartości na `0.1`, `0.2` oraz oczekiwaną sumę na `0.3`. Następnie znajdź inny przykład, dla którego porównanie daje `True`.


```python
a = 0.1
b = 0.1
oczekiwana_suma = 0.2
print("Suma:", a + b)
print("Czy równe?", a + b == oczekiwana_suma)
```

## 3. Wyświetlić ładnie czy porównać poprawnie?

### 3.1. Wygląd a wartość

Zapis `f"{wynik:.2f}"` tworzy tekst przedstawiający liczbę z dwoma miejscami po przecinku. Nie musisz teraz zapamiętywać jego składni.

**Przewidź:** czy po takim wyświetleniu porównanie z `0.3` stanie się prawdziwe?


```python
wynik = 0.1 + 0.25
print("Z dwoma miejscami:", f"{wynik:.2f}")
print("Czy równe 0.3?", wynik == 0.35)
print("Wartość nadal:", wynik)
```

### 3.2. Czy wyniki są dostatecznie bliskie?

`import math` udostępnia funkcje matematyczne. `math.isclose(a, b)` porównuje liczby z tolerancją zamiast wymagać identyczności.

Uruchom przykład:


```python
import math

wynik = 0.1 + 0.2
print("Dokładna równość:", wynik == 0.3)
print("Bliskość:", math.isclose(wynik, 0.3))
```

### 3.3. Tolerancja ma znaczenie

Sprawdzamy obliczoną długość względem wartości odniesienia. W tym zadaniu **umownie dopuszczamy różnicę 0.01 mm**.

- `abs_tol` to tolerancja bezwzględna, wyrażona w tych samych jednostkach co porównywane wartości.
- `rel_tol` to tolerancja względna. Ustawiamy ją tu na `0`, aby korzystać wyłącznie z progu bezwzględnego.

1. Przewidź wynik poniższego porównania i sprawdź go.
2. Zmień `10.005` na `10.02`. Czy porównanie nadal przechodzi?
3. Wróć do `10.005`. Zapisz obie długości w metrach: `0.010005` i `0.010`.
4. Zmień `abs_tol` aby zachować ten sam fizyczny próg.

```python
import math

wynik = 10.005       # mm
odniesienie = 10.0   # mm
tolerancja = 0.01    # mm

print("Różnica:", abs(wynik - odniesienie))
print("W granicach tolerancji:",
      math.isclose(wynik, odniesienie, rel_tol=0.0, abs_tol=tolerancja))
```

**Do zapamiętania:** formatowanie zmienia sposób prezentacji. Tolerancja określa kryterium porównania i wymaga uzasadnienia. Nie naprawia wcześniejszych obliczeń.

Domyślnie `math.isclose` używa tolerancji względnej `1e-9` i bezwzględnej `0.0`. Przy porównywaniu małej, niezerowej wartości z zerem zwykle trzeba jawnie określić odpowiednie `abs_tol`. Tolerancja względna skaluje się z większą wartością bezwzględną porównywanych liczb.

Błąd reprezentacji komputerowej oraz niepewność pomiarowa to różne rzeczy. Próg w powyższym zadaniu służy wyłącznie ćwiczeniu doboru tolerancji.

## 5. Mały argument, duża różnica: log i log1p

Chcemy obliczyć $\ln(1+x)$ dla małego dodatniego $x$.

W Pythonie:

- `math.log(y)` oblicza logarytm naturalny z `y`;
- `math.log1p(x)` oblicza $\ln(1+x)$ metodą zachowującą dokładność dla `x` bliskiego zeru.

Obie funkcje zwracają `float`. `log1p` nie zmienia typu liczby na bardziej precyzyjny - inaczej organizuje obliczenie.

### 5.1. Porównaj dwie drogi

Uruchom komórkę dla `x = 1e-8`. Następnie sprawdź `1e-12` i `1e-16`. Przy każdym kroku obejrzyj także wartość `1 + x`.


```python
import math

x = 1e-8
print("x:", x)
print("1 + x:", 1 + x)
print("log(1 + x):", math.log(1 + x))
print("log1p(x):", math.log1p(x))
```

### 5.3. Porównaj skalę rozbieżności

Poniżej dzielimy wynik przez `x`. Z przybliżenia wynika, że dla małego niezerowego `x` iloraz powinien być bliski 1.

Sprawdź `1e-8`, `1e-12` i `1e-16`.

Iloraz jest prostym sprawdzeniem sensowności wyniku, a nie obliczeniem dokładnego błędu względem wartości matematycznej. Samo przybliżenie $\ln(1+x)\approx x$ też ma błąd.


```python
import math

x = 1e-8
print("log(1 + x) / x:", math.log(1 + x) / x)
print("log1p(x) / x:", math.log1p(x) / x)
```

**Wniosek:** poprawny wzór matematyczny nie gwarantuje dokładnego wyniku w arytmetyce komputerowej. Znaczenie ma również sposób obliczenia i to, czy po drodze tracimy informację.

Podobnym przykładem funkcji do specjalnego zadania jest `math.expm1(x)`, obliczająca $e^x-1$ dokładnie dla `x` bliskiego zeru.
